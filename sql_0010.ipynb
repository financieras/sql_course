# 2. Cómo tomar tus apuntes

Una vez configurado, puedes usar el prefijo `%%sql` al principio de cualquier celda para que Jupyter se comporte como una consola de SQL:

> **Apunte: Creando mi primera tabla de clientes**
> Usamos `CREATE TABLE` para definir la estructura.
> ```sql

> CREATE TABLE estudiantes (

> );
>
> ```

````python
%%sql
CREATE TABLE estudiantes (
    id INTEGER PRIMARY KEY,
    nombre TEXT,
    nota DECIMAL
);
````
