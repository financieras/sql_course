## 3. Cómo tomar tus apuntes

Una vez configurado, puedes usar el prefijo `%%sql` al principio de cualquier celda para que Jupyter se comporte como una consola de SQL:

> Creando mi primera tabla de películas: movies**  
> - Usamos `CREATE TABLE` para definir la estructura.  
> - Es importante indicar el tipo de dato: TEXT, INTEGER, DECIMAL, ...

````python
%%sql
-- 1. Borramos la tabla si existe para limpiar datos viejos
DROP TABLE IF EXISTS movies;

-- 2. Creamos la tabla desde cero
CREATE TABLE movies (
    id INTEGER PRIMARY KEY,
    title TEXT,
    director TEXT,
    year INTEGER,
    length_minutes INTEGER
);
````

## 4. Inserción de los datos

Para meter todos tus datos de golpe, usamos el comando `INSERT INTO`. Fíjate que los textos van entre comillas simples.

*(Nota: En "A Bug's Life" he puesto doble comilla simple `''` porque en SQL es la forma de "escapar" un apóstrofe dentro de un texto).*

````python
%%sql
-- 3. Insertamos los datos
INSERT INTO movies (id, title, director, year, length_minutes) VALUES
(1, 'Toy Story', 'John Lasseter', 1995, 81),
(2, 'A Bug''s Life', 'John Lasseter', 1998, 95),
(3, 'Toy Story 2', 'John Lasseter', 1999, 93),
(4, 'Monsters, Inc.', 'Pete Docter', 2001, 92),
(5, 'Finding Nemo', 'Andrew Stanton', 2003, 107),
(6, 'The Incredibles', 'Brad Bird', 2004, 116),
(7, 'Cars', 'John Lasseter', 2006, 117),
(8, 'Ratatouille', 'Brad Bird', 2007, 115),
(9, 'WALL-E', 'Andrew Stanton', 2008, 104),
(10, 'Up', 'Pete Docter', 2009, 101),
(11, 'Toy Story 3', 'Lee Unkrich', 2010, 103),
(12, 'Cars 2', 'John Lasseter', 2011, 120),
(13, 'Brave', 'Brenda Chapman', 2012, 102),
(14, 'Monsters University', 'Dan Scanlon', 2013, 110);
````

## 5. Verificación

Para comprobar que todo está bien y ver tu tabla formateada en el cuaderno, ejecuta:

```sql
%%sql
SELECT * FROM movies;

```
